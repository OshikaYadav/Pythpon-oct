# ARITHMETIC OPERATORS #
The operators in which we can perform methametical operations.

In python there are 7 arithmetic operations.
These are:-
1. Addition (+)
2. Subtraction (-)
3. Multiplication (*)
4. Division (/)
5. Floor Division (//)
   These are used when the result of the number we want in whole number value not in the decimal value. 
6. Exponent (**)
   These are use to raise the power.
7. Modulus (%)
    These are used to get the answer of a division in its remainder value.

```python
#Arithmetic Operators on integers

num=100
num2 =3

#Adition
print(F'the addition of {num} and {num2} is{num+num2}')

#Subtraction
print(F'the subtraction of {num} and {num2} is {num-num2}')

#Multiplication
print(F'the multiplication of {num} and {num2} is {num*num2}')

#Division
print(F'the division of {num} and {num2} is {num/num2}')

#Floor Division
print(F'the floor division of {num} and {num2} is {num//num2}')

#Exponent
print(F'the exponent of {num} and {num2} is {num**num2}')

#Modulus
print(F'the modulus of {num} and {num2} is {num%num2}')
```

```python
#Arithmetic Operations on strings

msg1 = "Python"
msg2 = "Java"

#Addition
print(msg1+msg2)

#Additon Operations on string values gives result in string concatenation (joining of two strings)

#Multiplication
print(msg1*3)
# the string values only give the value of the two arithmetic operations only Addition nad Multiplication only.
# Both the never multiply and give the result but one string is being multiplpied with an integer value and give the result.
```

# COMPARISION OPERATORS #
Use to compare two values or variables.
The result of comparision operators is always true or fale.

## Comparision  operators in python ##
1. Greater than(>)
2. Less than (<)
3. Greater than equal to (>=)
4. Less than equal to (<=)
5. Not equal to (!=)
6. Equal to (==)

### single sign of = means value assign to the varable ###
### double sign of == means we are comparing two values whether they are equal or not ###

```python
#Comparision Operators :

# a. Greater than(>)
print(num>num2)

# b. Less than(<)
print(num<num2)

# c. Greater than equal to(>=)
print(num>=num2)

num3=100
num4 =100

# d. Less than equal to(<=)
print(num3<=num4)
print(num3>num4)

# e. Not equal to(!=)
print(num!=num2)
print(num3!=num4)

# f. Equal to (==)
print (num3==num4)
print(num==num2)
```
