# Tuple-
Tuple is an immutable , ordered collection of elements
## Characteristic of tuple
* 1. Ordered
* 2. Immutable
*  3. Allow duplicate va;ues
 * 4. Can contain different data types
  

```python
# Creating a tuple

marks =(12,23,45,15,11,14)
print(marks)

print(type(marks))
```

```python
# Creating a single element tuple

marks = (20)
print(marks)
print(type(marks))

# a single element inside round bracket is not condidered tuple, to create a single element tuple we have to follow this syntax:
marks = (20,)
print(marks)
print(type(marks))
```

```python
marks = (30)
print(marks)
print(type(marks))

marks(30,)
print(marks)
```

```python
# Indexing and slicing
names =('Alice','Charlie','John','Smith','Bob')
print(names[2])
print(names[-4])
```

```python
# tuple slicing
employee_targets = (2300,23400,4500,65000,23000,23210,31000,21310,3212321)

print(employee_targets[ : ])
print(employee_targets[ 0:8:2 ])
print(employee_targets[-1:-7:-2])
print(employee_targets[ -4: :2])
print(employee_targets[ : 3])
```

```python
# Tuple trevarsal
employee_targets = (2300,23400,4500,65000,23000,23210,31000,21310,3212321)
for target in employee_targets:
    print(target)
```

```python
# Checking membership operators
print(50000 in employee_targets)
print(45000 not in employee_targets)
```

```python
# in -built function for tuples
# sum()
print(sum(employee_targets))

# max()
print(max(employee_targets))

#min()
print(min(employee_targets))

#len()
print(len(employee_targets))

#sorted() - returns list of sorted values
print(sorted(employee_targets))

print(sorted(employee_targets,reverse = True))
```

```python
# methods of tuples
print(employee_targets)

#count()
print(employee_targets.count(234234))

# index()
print(employee_targets.index(23400))
# This tells about the position of the valued number.
```

```python
a = 87,34,23
print(a)
print(type(a))
# when assign multiple values to one variable then it forms tuple either with round brackets or not this is known as tuple packing.
```

```python
marks =(10,12,13,9,19)
(stu_1,stu_2,stu_3,stu_4,stu_5) = marks
print(stu_1)

# assigning multiple values into single values to different variable is called tuple unpacking.
```

```python
marks =(10,12,13,9,19)
(stu1,stu2,*stu) = marks
print(stu1,stu2,stu)
# This is called standard tuple unpacking.

(stu1,*stu,stu4,stu5) = marks 
print(stu1,stu,stu4,stu5)
```

```python
listt = [10,20,30,40,50]
new_tuple = tuple(listt)
print(new_tuple)
```

```python
# tuple concatenation

t1 = (10,20,30)
t2 = (40,50,60)
t3 = t1+t2
print(t3)
```

```python
marks= (10,20,32,23,34,35)
(var1,var2,var3,var4,var5,var6) = marks
print(var1,var2,var3,var4,var5,var6)
```

```python
(var1,var2,*var) = marks
print(var1)
print(var)
```
