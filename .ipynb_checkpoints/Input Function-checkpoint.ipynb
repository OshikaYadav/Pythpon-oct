# TypeCasting
The proces of converting the datatype of the variables from one another is called typecastingin Python.

Typecasting is of two types:-
1. Implicit Typecasting
2. Expilicit typecasting

# Implicit Typecasting :
The conversion of the datatypes values or variables performed by the Python interpeerter to prevent data loss is called Implicit typecasting.
Implicit typecasting is done without any intervention of the programmer.

# Explicit Typecasting :
The type of casting is done by the programmer manually using some inbuilt Python function such as int(),float(),bool(),str()etc.
