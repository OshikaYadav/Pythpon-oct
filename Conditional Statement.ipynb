# Conditional statements
Also known as decisionn making statements.

These are used to execute the different blocks of code depending on different condition.
In Python we have followiing coonditional statement available:-
1. If
2. Elseif
3. Elif
4. Nested if

In these conditions space should be left compulsory otherwise condition will not be applied.
And this will run as the normal command.
* IF Statement :- If statement is used to execute a block of code only when some condition is True.
Block of codes means an indented group of statements.

```python
num=int(input("enter anumber"))
#print("num is greater than 100")
if num>100:
  print("user have entered:",num)  
  print("number is greater than 100")   

print("hello")
    
```
