```python
#List slicing
marks = [43,23,67,55,34,65,78]
print(marks[0:3])
```

```python
# Membership operators in list
print(67 not in marks)
print(23 in  marks)
```

```python
for i in marks:
    print(i)
```

```python
marks = [43,23,67,55,34,65,78]
# Modifying the existing list columns

marks[1] = 13
print(marks)
```

```python
# append() - automatically insert an element at the end of list
marks.append(11)
print(marks)
```

```python
# extend () - used to insert multiple values at the end of list
marks.extend([10,11,12])
print(marks)
# use if want to insert multiple values at one time only.
```

```python
# removing elements from  a list
# remove() - remove the 1st occurence of the number
marks.remove(11)
print(marks)

#pop() - remove the last element
marks.pop()
print(marks)

#clear() - remove the complete list
marks.clear()
print(marks)
```

```python
# some other useful in-built list methods

marks=[11,44,45,76,89,43,23,45]

#len()
print(len(marks))

# max()
print(max(marks))

#min()
print(min(marks))

#sum()
print(sum(marks))

#sorted() - by default sorts the list in ascending order
print(sorted(marks))
print(sorted(marks,reverse = True))  # to sort in descending,with the list name also specify reverse = True
```

```python
# list comprehension 

english_marks =[12,23,43,12,34]
new_marks =[]

for i in english_marks:
    new_marks.append(i+2)

print(new_marks)
```
