# LOOP ( Control Flow Statements)
Loops are used to execute the block of code repeatedly until some condition is true or are generally used to ilterate over sequence.

In Pyhtonn we have two main loops:-
1. While loop - It is used to execute a code multiple number of times until specified condition remains true.
2. For loop - It is used to ilterate over seqences.
   

```python
corr_username = "admin"
input_username = input("Enter Username")

while input_username != corr_username:
    print("Incorrect Username.Try again!!!")
    input_username = input("Enter Username")

# If the condition is true then the coommand execute ,If the condition not rue or false then the loop exit.
# infinite loop exit that many times you enter tha false comdition.
```

```python
num = int(input("enter anumber"))
count = 0
total = 0
while (num >0):
    print("num entered:",num)
    count+=1
    total +=num
    num = int(input("enter a number"))
    
print("no of positive integer:",count)
print("sum of positive integer:",total)

# 1 se start kiya to sare no of times aayega lekin agar 0 se start karnege to uske aage ke number ko count karega
```
