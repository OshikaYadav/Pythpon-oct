```python
# In built methods that operate in strings:
#1. Case conversion methods
msg = "Python is easy"
#A. upper()
print(msg.upper())


#B. lower()
print(msg.lower())


#C. title() - it make sthe every word 1st letter in uppercase.
print(msg.title())

#D. swapcase() - it inverts the uppercase into lowercase and lowercase into uppercase.
print(msg.swapcase())

#E. capitalize() - it makes every sentence 1st letter in uppercase
print(msg.capitalize())
```

```python
#2. Removing extra spaces from string

msg = "      python    "
#A. strip()
print (msg.strip())


#B. lstrip()
print(msg.lstrip())

#c. rsteip()
print (msg.rstrip())
```

```python
# String validation methods

str1="Python"
str2 = "Pyhton123"
str3 = "PYTHON"
str4= "python is easy"
#A. isupper() - returns true if alll the alphabets in string are in upper case.
print(str1.isupper())
print(str3.isupper())


#B. islower() - returns true if all the value are in lower case.
print(str1.islower())
print (str4.islower())


#C. isalpha() - returns true if all the character in a string are alphabets.
print(str2.isalpha())
print(str1.isalpha())

#D. isalnum() - alum stands for alphanumeric , returns true if all the characters of a string are either alphabets or digits.
print(str1.isalnum())
print(str2.isalnum())
print(str4.isalnum())

#E. isspace() - returns true if string contains only space.
print(str3.isspace())
print(" ".isspace())

#F. isnumeric() - returns true if all the string elements are digits.
print(str3.isnumeric())
print("123".isnumeric())
```

```python
# 4. Searchig in strings
msg = "python is easy"

#A. find()
print (msg.find("y"))
print (msg.find("j"))

#B. index()
print (msg.index("y"))
print (msg.find("j"))

#Both find and index are ysed to check the index of some characters of multiple characters in a string..
# The difernece is when we try to find a chararcter thart is not avaible in a string, finds returns -1 and index raise error..

#C.count() - will return the no of occurences of the specified elements.
print (msg.count("y"))
print (msg.count("j"))

#D. startswith() - checks if string starts with specified characters or not.
print (msg.startswith("p"))

#E. endswith() - checks if string ends with specifiesd chracter or not.
print (msg.endswith("y"))
```

```python
# replace()
msg="python is easy"
print(msg.replace("y","Y"))

print(msg.replace("y","Y"),2)
```

```python
upper = 0
lower= 0
digits=0
spaces=0
string = input("enter a string")
for character in string:
    if character.isupper():
        upper +=1
    elif character.islower():
        lower +=1
    elif character.isnumeric():
        digits +=1
    elif character.isspace():
        spaces +=1

print ("total number of characters:",len(string))
print("uppercase charcters:",upper)
print("lowercase chracters:",lower)
print("digits:",digits)
print("spaces:",spaces)
```
